```python
import requests
import pandas as pd
import polars as pl
from datetime import datetime, timedelta

df = pd.read_json("evenements-publics-openagenda.json")


# ✨ CALCULER LA DATE IL Y A 1 AN
date_un_an = (datetime.now() - timedelta(days=365))

print(f"Date d'aujourd'hui : {datetime.now()}")
print(f"Date il y a 1 an : {date_un_an}")
print(f"\nAvant filtre : {len(df)} événements")

# ✨ FILTRER : Garder les événements >= il y a 1 an
df = df[df['lastdate_begin'] >= datetime.now()]

print(f"Après filtre : {len(df)} événements")
print(f"\nDate la plus ancienne : {df['lastdate_begin'].min()}")
print(f"Date la plus récente : {df['lastdate_begin'].max()}")
```
